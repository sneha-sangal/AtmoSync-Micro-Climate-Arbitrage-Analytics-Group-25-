```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
iot = pd.read_csv("../data/IoT_Sensor_Data_Cleaned.csv")
shipment = pd.read_csv("../data/Shipment_Details_Cleaned.csv")
market = pd.read_csv("../data/Market_Prices_Cleaned.csv")
route = pd.read_csv("../data/Route_Details_Cleaned.csv")
container = pd.read_csv("../data/Container_Details_Cleaned.csv")

print("AtmoSync datasets ready!")
```

```python
temperature_stats = iot["Temperature_C"].agg([
    "count",
    "mean",
    "median",
    "min",
    "max",
    "std"
])

print("TEMPERATURE STATISTICAL ANALYSIS")
display(temperature_stats)
```

```python
humidity_stats = iot["Humidity_Percent"].agg([
    "count",
    "mean",
    "median",
    "min",
    "max",
    "std"
])

print("HUMIDITY STATISTICAL ANALYSIS")
display(humidity_stats)
```

```python
vibration_stats = iot["Vibration_Level"].agg([
    "count",
    "mean",
    "median",
    "min",
    "max",
    "std"
])

print("VIBRATION STATISTICAL ANALYSIS")
display(vibration_stats)
```

```python
sensor_summary = iot[
    ["Temperature_C", "Humidity_Percent", "Vibration_Level"]
].agg(["mean", "median", "min", "max", "std"])

print("COMBINED SENSOR STATISTICAL SUMMARY")
display(sensor_summary.round(2))
```

```python
container_sensor_summary = iot.groupby("Container_ID")[
    ["Temperature_C", "Humidity_Percent", "Vibration_Level"]
].agg(["mean", "min", "max", "std"])

print("CONTAINER-WISE SENSOR ANALYSIS")
display(container_sensor_summary.head(10).round(2))
```

```python
print("TOTAL NUMBER OF CONTAINERS:", len(container_sensor_summary))

print("\nHIGHEST AVERAGE TEMPERATURE:")
display(
    container_sensor_summary[
        ("Temperature_C", "mean")
    ].sort_values(ascending=False).head(10).round(2)
)

print("\nLOWEST AVERAGE TEMPERATURE:")
display(
    container_sensor_summary[
        ("Temperature_C", "mean")
    ].sort_values().head(10).round(2)
)
```

```python
plt.figure(figsize=(15, 6))

container_sensor_summary[("Temperature_C", "mean")].plot(
    kind="bar"
)

plt.title("Average Temperature Across All Containers")
plt.xlabel("Container ID")
plt.ylabel("Average Temperature (°C)")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

print("Seaborn imported successfully!")
```

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 5))

sns.histplot(data=iot, x="Temperature_C", kde=True, ax=axes[0])
axes[0].set_title("Temperature Distribution")

sns.histplot(data=iot, x="Humidity_Percent", kde=True, ax=axes[1])
axes[1].set_title("Humidity Distribution")

sns.histplot(data=iot, x="Vibration_Level", kde=True, ax=axes[2])
axes[2].set_title("Vibration Distribution")

plt.tight_layout()
plt.show()
```
