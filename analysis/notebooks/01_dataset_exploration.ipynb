```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Pandas version:", pd.__version__)
print("NumPy version:", np.__version__)

print("AtmoSync Data Analysis Started Successfully!")
```

```python
iot = pd.read_csv("../data/IoT_Sensor_Data_Cleaned.csv")

shipment = pd.read_csv("../data/Shipment_Details_Cleaned.csv")

market = pd.read_csv("../data/Market_Prices_Cleaned.csv")

route = pd.read_csv("../data/Route_Details_Cleaned.csv")

container = pd.read_csv("../data/Container_Details_Cleaned.csv")

print("All 5 datasets loaded successfully!")
```

```python
datasets = {
    "IoT Sensor Data": iot,
    "Shipment Details": shipment,
    "Market Prices": market,
    "Route Details": route,
    "Container Details": container
}

for name, df in datasets.items():
    print(name, ":", df.shape)
```

```python
for name, df in datasets.items():

    print("\nDATASET:", name)
    print("-" * 50)

    print("Columns:")
    print(df.columns.tolist())

    print("\nData Types:")
    print(df.dtypes)

    print("\nFirst 5 Records:")
    display(df.head())
```

```python
for name, df in datasets.items():

    print("\n", name)
    print(df.isnull().sum())
```
